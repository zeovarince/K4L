# Eigen Vektor
## Definisi Eigen Vektor
*Definisi Eigen Vektor*

Eigenvector atau vektor eigen adalah vektor yang tidak berubah arah ketika diberikan transformasi linear. Dalam matematika, eigenvector sering digunakan dalam perhitungan matriks dan analisis spektral. Eigenvector juga dapat digunakan untuk memodelkan sistem dinamis dan mengidentifikasi pola dalam data. dapat dilihat dengan persamaan dibawah ini:

$$
A \mathbf{v} = \lambda \mathbf{v}
$$

Di mana:

A adalah matriks berukuran n × n.

v adalah vektor eigen yang ingin kita temukan.

λ adalah nilai eigen yang terkait dengan vektor eigen v.


## Mencari Eigen Vektor
Carilah Eigen Vektor dibawah ini dengan polinomal karakteristik:

$$A = \begin{bmatrix}
3 & 0 \\
0 & 3
\end{bmatrix}$$



$$B = \begin{bmatrix}
-2 & 0 \\
0 & 4
\end{bmatrix}$$

jawab:

Vektor Eigen Matriks $ A $

Dari perhitungan sebelumnya, kita mendapatkan nilai eigen dari $ A $:
$
\lambda = 3 \quad (\text{dengan multiplicity 2})
$

Untuk mencari vektor eigen, kita perlu menyelesaikan persamaan:
$
(A - \lambda I) \mathbf{v} = 0
$

Dengan $ \lambda = 3 $:
$
A - 3I = \begin{bmatrix}
3 - 3 & 0 \\
0 & 3 - 3
\end{bmatrix} = \begin{bmatrix}
0 & 0 \\
0 & 0
\end{bmatrix}
$

Persamaan ini menjadi:
$
\begin{bmatrix}
0 & 0 \\
0 & 0
\end{bmatrix} \begin{bmatrix}
x_1 \\
x_2
\end{bmatrix} = \begin{bmatrix}
0 \\
0
\end{bmatrix}
$

Ini menunjukkan bahwa setiap vektor dalam bentuk:
$
\mathbf{v} = \begin{bmatrix}
x_1 \\
x_2
\end{bmatrix} = \begin{bmatrix}
1 \\
0
\end{bmatrix} \text{ dan } \begin{bmatrix}
0 \\
1
\end{bmatrix}
$
adalah vektor eigen yang sesuai dengan nilai eigen $ \lambda = 3 $.

Jadi, vektor eigen untuk matriks $ A $ adalah:
$
\boxed{
\begin{bmatrix}
1 \\
0
\end{bmatrix}, \quad
\begin{bmatrix}
0 \\
1
\end{bmatrix}
}
$

\subsection*{Vektor Eigen Matriks $ B $}

Dari perhitungan sebelumnya, kita mendapatkan nilai eigen dari $ B $:

$$\lambda_1 = -2, \quad \lambda_2 = 4$$

\textbf{Untuk } $ \lambda_1 = -2 $:


$$B - (-2)I = B + 2I = \begin{bmatrix}
-2 + 2 & 0 \\
0 & 4 + 2
\end{bmatrix} = \begin{bmatrix}
0 & 0 \\
0 & 6
\end{bmatrix}$$

Persamaan ini menjadi:

$$\begin{bmatrix}
0 & 0 \\
0 & 6
\end{bmatrix} \begin{bmatrix}
x_1 \\
x_2
\end{bmatrix} = \begin{bmatrix}
0 \\
0
\end{bmatrix}$

Ini menunjukkan bahwa setiap vektor dalam bentuk:

$$\mathbf{v} = \begin{bmatrix}
x_1 \\
x_2
\end{bmatrix} = \begin{bmatrix}
1 \\
0
\end{bmatrix}$$
adalah vektor eigen yang sesuai dengan nilai eigen $ \lambda_1 = -2 $.

Jadi, vektor eigen untuk $ \lambda_1 = -2 $ adalah:

$$\boxed{
\begin{bmatrix}
1 \\
0
\end{bmatrix}
}$$

\textbf{Untuk } $ \lambda_2 = 4 $:


$$B - 4I = \begin{bmatrix}
-2 - 4 & 0 \\
0 & 4 - 4
\end{bmatrix} = \begin{bmatrix}
-6 & 0 \\
0 & 0
\end{bmatrix}$$

Persamaan ini menjadi:

$$\begin{bmatrix}
-6 & 0 \\
0 & 0
\end{bmatrix} \begin{bmatrix}
x_1 \\
x_2
\end{bmatrix} = \begin{bmatrix}
0 \\
0
\end{bmatrix}$$

Ini menunjukkan bahwa setiap vektor dalam bentuk:

$$\mathbf{v} = \begin{bmatrix}
0 \\
x_2
\end{bmatrix} = \begin{bmatrix}
0 \\
1
\end{bmatrix}$$
adalah vektor eigen yang sesuai dengan nilai eigen $ \lambda_2 = 4 $.

Jadi, vektor eigen untuk $ \lambda_2 = 4 $ adalah:

$$\begin{bmatrix}
0 \\
1
\end{bmatrix}$$

Kesimpulan

Vektor eigen untuk matriks $ A $ dan $ B $ adalah:
$$
\text{Matriks } A: \quad \begin{bmatrix}
1 \\
0
\end{bmatrix}, \quad \begin{bmatrix}
0 \\
1
\end{bmatrix}$$

$$\text{Matriks } B: \quad \begin{bmatrix}
1 \\
0
\end{bmatrix}, \quad \begin{bmatrix}
0 \\
1
\end{bmatrix}$$

```python
import numpy as np

# Definisikan matriks A dan B
A = np.array([[3, 0],
              [0, 3]])

B = np.array([[-2, 0],
              [0, 4]])

# Hitung nilai eigen dan vektor eigen matriks A
eigenvalues_A, eigenvectors_A = np.linalg.eig(A)
print("Nilai eigen matriks A:", eigenvalues_A)
print("Vektor eigen matriks A:")
print(eigenvectors_A)

# Hitung nilai eigen dan vektor eigen matriks B
eigenvalues_B, eigenvectors_B = np.linalg.eig(B)
print("\nNilai eigen matriks B:", eigenvalues_B)
print("Vektor eigen matriks B:")
print(eigenvectors_B)
```

## Keterkaitan antara Eigenvalue dan Eigenvector

Dari matriks A dan B yang diberikan kita dapat menyimpulkan keterkaitan antara eigenvalues dan eigenvektor salah satunya adalah hubungan matematis yang dimana Setiap nilai eigen λ memiliki vektor eigen v yang terkait. Ketika kita menemukan nilai eigen dari matriks, kita dapat menggunakan nilai tersebut untuk menemukan vektor eigen yang sesuai. Dalam konteks matriks A dan B:

Untuk matriks A, kita menemukan nilai eigen λ = 3 dengan multiplicity 2. Vektor eigen yang terkait adalah [1 0] dan [0 1], yang menunjukkan bahwa setiap vektor dalam ruang vektor dua dimensi akan tetap berada di sepanjang sumbu yang sama setelah transformasi oleh matriks A.

Untuk matriks B, kita menemukan dua nilai eigen: λ₁ = –2 dan λ₂ = 4. Vektor eigen yang terkait dengan λ₁ adalah [1 0] dan untuk λ₂ adalah [0 1]. Ini menunjukkan bahwa vektor eigen untuk λ₁ akan dibalik arah (karena negatif) dan vektor eigen untuk λ₂ akan diperbesar.
