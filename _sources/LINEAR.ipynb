# Transformasi Linear

## Definisi

Transformasi linear adalah sebuah fungsi yang memetakan suatu ruang vektor ke ruang vektor lainnya sedemikian rupa sehingga operasi penjumlahan vektor dan perkalian vektor dengan skalar tetap berlaku. Dalam konteks ini, jika 𝑇 adalah sebuah transformasi linier, maka untuk setiap vektor 𝑢 dan 𝑣 serta skalar berlaku:

## Sifat-sifat Transformasi Linear


### 1. Additivitas (penjumlahan vektor):

$$ 𝑇(𝑢+𝑣)=𝑇(𝑢)+𝑇(𝑣)T(u+v)=T(u)+T(v) $$

### 2. Homogenitas skalar (perkalian dengan skalar):

$$𝑇(𝑐𝑣)=𝑐𝑇(𝑣)$$

Untuk setiap skalar 𝑐 dan vektor 𝑣 di domainnya.
Dengan kata lain, transformasi linear melestarikan struktur linear dari ruang vektor, artinya hasil transformasi dari kombinasi linear vektor tetap merupakan kombinasi linear dari hasil transformasi vektor-vektor tersebut.

## Contoh Additivitas :
###Contoh Transformasi Linear dari $(\mathbb{R}^2)$  ke  $(\mathbb{R}^2)$


Misalkan kita definisikan transformasi linear:

$$T : \mathbb{R}^2 \rightarrow \mathbb{R}^2$$

$$T(x, y) = (2x + y,\ x - y)$$

Ambil dua vektor dari $(\mathbb{R}^2)$:


$$\mathbf{u} = (1, 2),\quad \mathbf{v} = (3, -1)$$


$$\mathbf{u} + \mathbf{v} = (1 + 3,\ 2 + (-1)) = (4, 1)$$


$$T(4, 1) = (2 \cdot 4 + 1,\ 4 - 1) = (9, 3)$$

$$T(1, 2) = (2 \cdot 1 + 2,\ 1 - 2) = (4, -1)$$

$$T(3, -1) = (2 \cdot 3 + (-1),\ 3 - (-1)) = (5, 4)$$


$$T(\mathbf{u}) + T(\mathbf{v}) = (4, -1) + (5, 4) = (9, 3)$$


###Contoh Transformasi Linear dari $(\mathbb{R}^2)$  ke  $(\mathbb{R}^3)$

Misalkan kita definisikan transformasi linear:


$$T : \mathbb{R}^2 \rightarrow \mathbb{R}^3$$

$$T(x, y) = (x + y,\ 2x,\ -y)$$

Ambil dua vektor dari $(\mathbb{R}^2):$

$$\mathbf{u} = (1, 2),\quad \mathbf{v} = (3, -1)$$

$$\mathbf{u} + \mathbf{v} = (1+3,\ 2 + (-1)) = (4, 1)$$

$$T(4, 1) = (4 + 1,\ 2 \cdot 4,\ -1) = (5, 8, -1)$$

$$T(1, 2) = (1 + 2,\ 2 \cdot 1,\ -2) = (3, 2, -2)$$

$$T(3, -1) = (3 + (-1),\ 2 \cdot 3,\ 1) = (2, 6, 1)$$

$$T(\mathbf{u}) + T(\mathbf{v}) = (3, 2, -2) + (2, 6, 1) = (5, 8, -1)$$


## Contoh Homogenitas skalar (perkalian dengan skalar):


Misalkan kita definisikan transformasi linear sebagai berikut:


$$T : \mathbb{R}^2 \rightarrow \mathbb{R}^2$$

$$T(x, y) = (2x + y,\ x - y)$$

Ambil vektor dari $(\mathbb{R}^2)$:


$$\mathbf{v} = (1, 3)$$

Dan ambil skalar $$(c = 4).$$


$$c \mathbf{v} = 4 \cdot (1, 3) = (4, 12)$$


$$T(4, 12) = (2 \cdot 4 + 12,\ 4 - 12) = (8 + 12,\ -8) = (20,\ -8)$$


$$T(1, 3) = (2 \cdot 1 + 3,\ 1 - 3) = (5,\ -2)$$

$$c \cdot T(\mathbf{v}) = 4 \cdot (5,\ -2) = (20,\ -8)$$

Kesimpulan:

$$T(c \mathbf{v}) = c \cdot T(\mathbf{v})$$

Maka terbukti bahwa transformasi \(T\) memenuhi sifat homogenitas skalar.


## Tugas

Tunjukkan bahwa transformasi linear

$$T : \mathbb{R}^2 \rightarrow \mathbb{R}^2$$

dinyatakan dengan matriks

$$A = \begin{bmatrix}
\cos \theta & -\sin \theta \\
\sin \theta & \cos \theta
\end{bmatrix}$$

memiliki sifat rotasi dalam $(\mathbb{R}^2)$ berlawanan dengan arah jarum jam sebesar sudut $(\theta)$.

Solusi:

Pembuktian Matriks Rotasi

Diberikan transformasi linear $( T : \mathbb{R}^2 \to \mathbb{R}^2 )$ dengan matriks:


$$A = \begin{bmatrix}
\cos \theta & -\sin \theta \\
\sin \theta & \cos \theta
\end{bmatrix}$$

Kita ingin membuktikan bahwa $(A)$ merepresentasikan rotasi berlawanan arah jarum jam sebesar sudut $(\theta)$.
Langkah 1: Representasi Vektor dalam Koordinat Polar

Misalkan vektor $$(\mathbf{v} = (x, y))$$ dinyatakan dalam koordinat polar sebagai:

$$\mathbf{v} = (x, y) = (r \cos \alpha, r \sin \alpha)$$

Langkah 2: Kalikan Matriks Rotasi dengan Vektor

Terapkan transformasi $(T(\mathbf{v}) = A \mathbf{v}):$


$$T(\mathbf{v}) =
\begin{bmatrix}
\cos \theta & -\sin \theta \\
\sin \theta & \cos \theta
\end{bmatrix}
\begin{bmatrix}
r \cos \alpha \\
r \sin \alpha
\end{bmatrix}
=
\begin{bmatrix}
r \cos \theta \cos \alpha - r \sin \theta \sin \alpha \\
r \sin \theta \cos \alpha + r \cos \theta \sin \alpha
\end{bmatrix}$$

Gunakan identitas trigonometri:


$$\cos(\alpha + \theta) = \cos \alpha \cos \theta - \sin \alpha \sin \theta$$

$$\sin(\alpha + \theta) = \sin \alpha \cos \theta + \cos \alpha \sin \theta$$

Sehingga:


$$T(\mathbf{v}) =
\begin{bmatrix}
r \cos(\alpha + \theta) \\
r \sin(\alpha + \theta)
\end{bmatrix}$$

Kesimpulan

Hasil transformasi $(T(\mathbf{v}))$ adalah vektor dengan panjang tetap $(r)$, tetapi sudutnya bertambah menjadi $(\alpha + \theta)$.  
Artinya, transformasi ini memutar vektor $(\mathbf{v})$ sebesar sudut $(\theta)$ berlawanan arah jarum jam.

$$T(\mathbf{v}) = A \mathbf{v} \text{ adalah rotasi sebesar } \theta \text{ terhadap titik asal}$$
