# Dot Product

## Pengertian Dot Product

>Perkalian titik atau dot product dua buah vektor didefinisikan sebagai perkalian antara besar salah satu vektor (misalA) dengan komponen vektor kedua (B) pada arah vektor pertama (A). Pada gambar di atas, komponen vektor B pada arah vektor A adalah B cos α. Jika kita memiliki dua vektor $\mathbf{a} = (a_1, a_2, \ldots, a_n)$ dan $\mathbf{b} = (b_1, b_2, \ldots, b_n)$, maka dot product dari $\mathbf{a}$ dan $\mathbf{b}$ didefinisikan sebagai:

$$
\mathbf{a} \cdot \mathbf{b} = a_1 b_1 + a_2 b_2 + \ldots + a_n b_n
$$



>Secara geometris, dot product juga dapat dinyatakan dalam bentuk:
$$
\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \cos(\theta)
$$

## Contoh Perkalian Dot

Diberikan dua vektor:
$$
\mathbf{a} = (3, \,-2, \,5) \quad \text{dan} \quad \mathbf{b} = (4, \,0, \,-1)
$$

Hitunglah $$\mathbf{a} \cdot \mathbf{b}?$$

Penyelesaian:

Dot product dari \$\mathbf{a}\$ dan \$\mathbf{b}\$ didefinisikan sebagai:

$$
\mathbf{a} \cdot \mathbf{b} = a_1 b_1 + a_2 b_2 + a_3 b_3
$$

Menghitung setiap komponen:
$$
3 \times 4 = 12, \quad (-2) \times 0 = 0, \quad 5 \times (-1) = -5
$$

Jadi,
$$
\mathbf{a} \cdot \mathbf{b} = 12 + 0 - 5 = 7
$$

Jadi, hasil dot product $$\mathbf{a} \cdot \mathbf{b} = 7$$




## Code Program Mencari Perkalian Dor

```python
import numpy as np

# Definisikan vektor
a = np.array([3, -2, 5])
b = np.array([4, 0, -1])

# Hitung dot product
hasil = np.dot(a, b)

print(a,b)
print("Hasil dot product:", hasil)
```

## Vektor Satuan

> Vektor satuan $\mathbf{u}$ didefinisikan sebagai vektor yang memenuhi syarat:

$$ ||\mathbf{u}|| = 1 $$

> di mana $||\mathbf{u}||$ adalah magnitudo (panjang) dari vektor $\mathbf{u}$.






## Menghitung Vektor Satuan

> Untuk mengubah vektor $\mathbf{v} = (x, y, z)$ menjadi vektor satuan, kita perlu menghitung magnitudo vektor tersebut terlebih dahulu:


$$ ||\mathbf{v}|| = \sqrt{x^2 + y^2 + z^2} $$


> Setelah mendapatkan magnitudo, vektor satuan $\mathbf{u}$ dapat dihitung dengan rumus:

$$ \mathbf{u} = \frac{\mathbf{v}}{||\mathbf{v}||} = \left( \frac{x}{||\mathbf{v}||}, \frac{y}{||\mathbf{v}||}, \frac{z}{||\mathbf{v}||} \right) $$

## Code Program

```python
import math

def vektor_satuan(v):
    # Menghitung magnitudo vektor
    magnitudo = math.sqrt(sum(comp**2 for comp in v))

    # Menghindari pembagian dengan nol
    if magnitudo == 0:
        raise ValueError("Vektor tidak boleh nol.")

    # Menghitung vektor satuan
    vektor_unit = [comp / magnitudo for comp in v]
    return vektor_unit

# Contoh penggunaan
vektor = [3, 4, 0]
vektor_unit = vektor_satuan(vektor)
print("Vektor Satuan:", vektor_unit)
```
